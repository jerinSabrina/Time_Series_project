```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# =========================
# EDA STEP 2: MISSING FRACTION DATES + NORMALIZATION CHECKS
# =========================

# Make sure dates are in datetime format
input_df["Zeitstempel"] = pd.to_datetime(input_df["Zeitstempel"])
fraction_df["Datum"] = pd.to_datetime(fraction_df["Datum"])

# Fraction columns only
fraction_cols = [col for col in fraction_df.columns if col not in ["Datum", "fraction_sum"]]

# Rename date columns for joining
input_join = input_df.rename(columns={"Zeitstempel": "date"}).copy()
fraction_join = fraction_df.rename(columns={"Datum": "date"}).copy()

# Outer join to see which dates are missing in fraction data
outer_joined = pd.merge(
    input_join,
    fraction_join,
    on="date",
    how="outer",
    indicator=True
)

outer_joined = outer_joined.sort_values("date").reset_index(drop=True)

print("===== OUTER JOIN STATUS =====")
display(outer_joined["_merge"].value_counts().to_frame("count"))
```

```python
# =========================
# 1. SPLIT LEFT_ONLY DATES
# =========================

fraction_start_date = fraction_df["Datum"].min()

left_only_dates = outer_joined[outer_joined["_merge"] == "left_only"].copy()

left_only_before_fraction_start = left_only_dates[
    left_only_dates["date"] < fraction_start_date
].copy()

true_missing_fraction_dates = left_only_dates[
    left_only_dates["date"] >= fraction_start_date
].copy()

missing_split_summary = pd.DataFrame({
    "category": [
        "left_only before fraction data starts",
        "left_only within fraction data period"
    ],
    "count": [
        len(left_only_before_fraction_start),
        len(true_missing_fraction_dates)
    ]
})

print("===== LEFT_ONLY DATE SPLIT =====")
display(missing_split_summary)

print("Fraction data starts on:", fraction_start_date)
```

```python
# =========================
# 2. MASSE STATUS ON TRUE MISSING FRACTION DATES
# =========================

missing_masse_summary = pd.DataFrame({
    "condition": [
        "Masse = 0",
        "Masse > 0",
        "Masse < 0"
    ],
    "count": [
        (true_missing_fraction_dates["Masse"] == 0).sum(),
        (true_missing_fraction_dates["Masse"] > 0).sum(),
        (true_missing_fraction_dates["Masse"] < 0).sum()
    ]
})

print("===== MASSE STATUS ON TRUE MISSING FRACTION DATES =====")
display(missing_masse_summary)

print("First true missing fraction dates:")
display(true_missing_fraction_dates[["date", "Masse"]].head(30))
```

```python
# =========================
# 3. WEEKDAY PATTERN OF TRUE MISSING FRACTION DATES
# =========================

true_missing_fraction_dates["weekday"] = true_missing_fraction_dates["date"].dt.day_name()
true_missing_fraction_dates["month"] = true_missing_fraction_dates["date"].dt.month
true_missing_fraction_dates["year"] = true_missing_fraction_dates["date"].dt.year

weekday_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]

missing_weekday_counts = (
    true_missing_fraction_dates["weekday"]
    .value_counts()
    .reindex(weekday_order)
    .fillna(0)
    .astype(int)
    .to_frame("missing_fraction_dates")
)

print("===== TRUE MISSING FRACTION DATES BY WEEKDAY =====")
display(missing_weekday_counts)

missing_weekday_counts.plot(kind="bar", figsize=(10, 4), legend=False)
plt.title("True Missing Fraction Dates by Weekday")
plt.xlabel("Weekday")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
# =========================
# 4. MONTHLY / YEARLY PATTERN OF TRUE MISSING FRACTION DATES
# =========================

missing_by_year_month = (
    true_missing_fraction_dates
    .groupby(["year", "month"])
    .size()
    .reset_index(name="missing_count")
)

print("===== TRUE MISSING FRACTION DATES BY YEAR-MONTH =====")
display(missing_by_year_month)

pivot_missing_month = missing_by_year_month.pivot(
    index="year",
    columns="month",
    values="missing_count"
).fillna(0)

display(pivot_missing_month)

plt.figure(figsize=(12, 4))
plt.imshow(pivot_missing_month, aspect="auto")
plt.colorbar(label="Missing fraction date count")
plt.title("Heatmap: Missing Fraction Dates by Year and Month")
plt.xlabel("Month")
plt.ylabel("Year")
plt.xticks(
    ticks=np.arange(len(pivot_missing_month.columns)),
    labels=pivot_missing_month.columns
)
plt.yticks(
    ticks=np.arange(len(pivot_missing_month.index)),
    labels=pivot_missing_month.index
)
plt.show()
```

```python
# =========================
# 5. NORMALIZATION CHECK: MASSE
# =========================

masse_normalization_check = pd.DataFrame({
    "metric": [
        "min",
        "max",
        "mean",
        "negative_count",
        "zero_count",
        "greater_than_1_count",
        "between_0_and_1_count"
    ],
    "value": [
        input_df["Masse"].min(),
        input_df["Masse"].max(),
        input_df["Masse"].mean(),
        (input_df["Masse"] < 0).sum(),
        (input_df["Masse"] == 0).sum(),
        (input_df["Masse"] > 1).sum(),
        ((input_df["Masse"] >= 0) & (input_df["Masse"] <= 1)).sum()
    ]
})

print("===== MASSE NORMALIZATION CHECK =====")
display(masse_normalization_check)
```

```python
# =========================
# 6. NORMALIZATION CHECK: FRACTIONS
# =========================

fraction_df["fraction_sum"] = fraction_df[fraction_cols].sum(axis=1)

fraction_sum_check = pd.DataFrame({
    "metric": [
        "min_fraction_sum",
        "max_fraction_sum",
        "mean_fraction_sum",
        "rows_sum_close_to_1",
        "rows_sum_not_close_to_1"
    ],
    "value": [
        fraction_df["fraction_sum"].min(),
        fraction_df["fraction_sum"].max(),
        fraction_df["fraction_sum"].mean(),
        np.isclose(fraction_df["fraction_sum"], 1.0).sum(),
        (~np.isclose(fraction_df["fraction_sum"], 1.0)).sum()
    ]
})

print("===== FRACTION NORMALIZATION CHECK =====")
display(fraction_sum_check)
```

```python
# =========================
# 7. FRACTION ZERO PATTERN
# =========================

fraction_zero_summary = pd.DataFrame({
    "zero_count": (fraction_df[fraction_cols] == 0).sum(),
    "zero_percentage": (fraction_df[fraction_cols] == 0).mean() * 100,
    "mean_fraction": fraction_df[fraction_cols].mean(),
    "max_fraction": fraction_df[fraction_cols].max()
}).sort_values("zero_percentage", ascending=False)

print("===== FRACTION ZERO / SPARSITY SUMMARY =====")
display(fraction_zero_summary)
```

```python
# =========================
# 8. HEATMAP: FRACTION VALUES OVER TIME
# =========================

fraction_heatmap_data = fraction_df.sort_values("Datum")[fraction_cols].T

plt.figure(figsize=(14, 5))
plt.imshow(fraction_heatmap_data, aspect="auto")
plt.colorbar(label="Fraction value")
plt.title("Heatmap: Fraction Values Over Time")
plt.xlabel("Time index")
plt.ylabel("Fraction category")
plt.yticks(
    ticks=np.arange(len(fraction_cols)),
    labels=fraction_cols
)
plt.show()
```

```python
# =========================
# 9. HEATMAP: ZERO PATTERN IN FRACTIONS
# =========================

zero_heatmap_data = (fraction_df.sort_values("Datum")[fraction_cols] == 0).T

plt.figure(figsize=(14, 5))
plt.imshow(zero_heatmap_data, aspect="auto")
plt.colorbar(label="Zero indicator")
plt.title("Heatmap: Zero Pattern in Fraction Columns Over Time")
plt.xlabel("Time index")
plt.ylabel("Fraction category")
plt.yticks(
    ticks=np.arange(len(fraction_cols)),
    labels=fraction_cols
)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# =========================
# EDA STEP 3: INPUT MASS ANALYSIS
# =========================

mass_df = input_df.copy()
mass_df["Zeitstempel"] = pd.to_datetime(mass_df["Zeitstempel"])
mass_df = mass_df.sort_values("Zeitstempel").reset_index(drop=True)

# Calendar features for EDA only
mass_df["year"] = mass_df["Zeitstempel"].dt.year
mass_df["month"] = mass_df["Zeitstempel"].dt.month
mass_df["weekday"] = mass_df["Zeitstempel"].dt.day_name()
mass_df["day_of_week"] = mass_df["Zeitstempel"].dt.dayofweek
mass_df["week"] = mass_df["Zeitstempel"].dt.isocalendar().week.astype(int)

weekday_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
```

```python
# =========================
# 1. BASIC MASSE SUMMARY
# =========================

print("===== MASSE BASIC SUMMARY =====")
display(mass_df["Masse"].describe().to_frame())

mass_quality_summary = pd.DataFrame({
    "condition": [
        "Masse < 0",
        "Masse = 0",
        "0 < Masse <= 1",
        "Masse > 1"
    ],
    "count": [
        (mass_df["Masse"] < 0).sum(),
        (mass_df["Masse"] == 0).sum(),
        ((mass_df["Masse"] > 0) & (mass_df["Masse"] <= 1)).sum(),
        (mass_df["Masse"] > 1).sum()
    ]
})

print("===== MASSE QUALITY SUMMARY =====")
display(mass_quality_summary)
```

```python
# =========================
# 2. NEGATIVE AND ZERO MASS DATES
# =========================

print("===== NEGATIVE MASSE DATES =====")
display(mass_df[mass_df["Masse"] < 0][["Zeitstempel", "Masse", "weekday"]])

print("===== FIRST ZERO MASSE DATES =====")
display(mass_df[mass_df["Masse"] == 0][["Zeitstempel", "Masse", "weekday"]].head(30))
```

```python
# =========================
# 3. TIME SERIES PLOT
# =========================

plt.figure(figsize=(14, 4))
plt.plot(mass_df["Zeitstempel"], mass_df["Masse"])
plt.title("Input Mass Over Time")
plt.xlabel("Date")
plt.ylabel("Normalized Masse")
plt.show()
```

```python
# =========================
# 4. ROLLING MEAN PLOTS
# =========================

mass_df["rolling_7"] = mass_df["Masse"].rolling(window=7, min_periods=1).mean()
mass_df["rolling_30"] = mass_df["Masse"].rolling(window=30, min_periods=1).mean()

plt.figure(figsize=(14, 4))
plt.plot(mass_df["Zeitstempel"], mass_df["Masse"], alpha=0.4, label="Daily Masse")
plt.plot(mass_df["Zeitstempel"], mass_df["rolling_7"], label="7-day rolling mean")
plt.plot(mass_df["Zeitstempel"], mass_df["rolling_30"], label="30-day rolling mean")
plt.title("Input Mass With Rolling Means")
plt.xlabel("Date")
plt.ylabel("Normalized Masse")
plt.legend()
plt.show()
```

```python
# =========================
# 5. DISTRIBUTION OF MASSE
# =========================

plt.figure(figsize=(8, 4))
plt.hist(mass_df["Masse"], bins=50)
plt.title("Distribution of Input Mass")
plt.xlabel("Normalized Masse")
plt.ylabel("Frequency")
plt.show()
```

```python
# =========================
# 6. WEEKDAY PATTERN
# =========================

weekday_mass_summary = (
    mass_df
    .groupby("weekday")["Masse"]
    .agg(["count", "mean", "median", "std", "min", "max"])
    .reindex(weekday_order)
)

print("===== MASSE BY WEEKDAY =====")
display(weekday_mass_summary)

plt.figure(figsize=(10, 4))
weekday_mass_summary["mean"].plot(kind="bar")
plt.title("Mean Input Mass by Weekday")
plt.xlabel("Weekday")
plt.ylabel("Mean Normalized Masse")
plt.xticks(rotation=45)
plt.show()
```

```python
# =========================
# 7. ZERO MASSE BY WEEKDAY
# =========================

zero_by_weekday = (
    mass_df.assign(is_zero_mass = mass_df["Masse"] == 0)
    .groupby("weekday")["is_zero_mass"]
    .agg(["sum", "mean"])
    .reindex(weekday_order)
)

zero_by_weekday["zero_percentage"] = zero_by_weekday["mean"] * 100

print("===== ZERO MASSE BY WEEKDAY =====")
display(zero_by_weekday[["sum", "zero_percentage"]])

plt.figure(figsize=(10, 4))
zero_by_weekday["zero_percentage"].plot(kind="bar")
plt.title("Percentage of Zero-Mass Days by Weekday")
plt.xlabel("Weekday")
plt.ylabel("Zero-Mass Percentage")
plt.xticks(rotation=45)
plt.show()
```

```python
# =========================
# 8. MONTHLY PATTERN
# =========================

monthly_mass_summary = (
    mass_df
    .groupby(["year", "month"])["Masse"]
    .agg(["count", "mean", "median", "std", "min", "max"])
    .reset_index()
)

print("===== MASSE BY YEAR-MONTH =====")
display(monthly_mass_summary)

monthly_pivot = monthly_mass_summary.pivot(
    index="year",
    columns="month",
    values="mean"
)

display(monthly_pivot)

plt.figure(figsize=(12, 4))
plt.imshow(monthly_pivot, aspect="auto")
plt.colorbar(label="Mean normalized Masse")
plt.title("Heatmap: Mean Input Mass by Year and Month")
plt.xlabel("Month")
plt.ylabel("Year")
plt.xticks(
    ticks=np.arange(len(monthly_pivot.columns)),
    labels=monthly_pivot.columns
)
plt.yticks(
    ticks=np.arange(len(monthly_pivot.index)),
    labels=monthly_pivot.index
)
plt.show()
```

```python
# =========================
# 9. YEARLY SUMMARY
# =========================

yearly_mass_summary = (
    mass_df
    .groupby("year")["Masse"]
    .agg(["count", "mean", "median", "std", "min", "max"])
)

print("===== MASSE BY YEAR =====")
display(yearly_mass_summary)

plt.figure(figsize=(8, 4))
yearly_mass_summary["mean"].plot(kind="bar")
plt.title("Mean Input Mass by Year")
plt.xlabel("Year")
plt.ylabel("Mean Normalized Masse")
plt.xticks(rotation=0)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# =========================
# EDA STEP 4: FRACTION / MATERIAL COMPOSITION ANALYSIS
# =========================

frac_df = fraction_df.copy()
frac_df["Datum"] = pd.to_datetime(frac_df["Datum"])
frac_df = frac_df.sort_values("Datum").reset_index(drop=True)

# Use only real fraction columns
fraction_cols = [col for col in frac_df.columns if col not in ["Datum", "fraction_sum"]]

# Calendar variables for EDA
frac_df["year"] = frac_df["Datum"].dt.year
frac_df["month"] = frac_df["Datum"].dt.month
frac_df["weekday"] = frac_df["Datum"].dt.day_name()

weekday_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
```

```python
# =========================
# 1. BASIC FRACTION SUMMARY
# =========================

print("===== FRACTION BASIC SUMMARY =====")

fraction_summary = pd.DataFrame({
    "mean": frac_df[fraction_cols].mean(),
    "median": frac_df[fraction_cols].median(),
    "std": frac_df[fraction_cols].std(),
    "min": frac_df[fraction_cols].min(),
    "max": frac_df[fraction_cols].max(),
    "zero_count": (frac_df[fraction_cols] == 0).sum(),
    "zero_percentage": (frac_df[fraction_cols] == 0).mean() * 100
}).sort_values("mean", ascending=False)

display(fraction_summary)
```

```python
# =========================
# 2. MEAN FRACTION SHARE BAR PLOT
# =========================

plt.figure(figsize=(10, 4))
fraction_summary["mean"].plot(kind="bar")
plt.title("Average Share of Each Material Fraction")
plt.xlabel("Fraction Category")
plt.ylabel("Mean Fraction")
plt.xticks(rotation=0)
plt.show()
```

```python
# =========================
# 3. ZERO / SPARSITY BAR PLOT
# =========================

plt.figure(figsize=(10, 4))
fraction_summary.sort_values("zero_percentage", ascending=False)["zero_percentage"].plot(kind="bar")
plt.title("Zero Percentage of Each Material Fraction")
plt.xlabel("Fraction Category")
plt.ylabel("Zero Percentage")
plt.xticks(rotation=0)
plt.show()
```

```python
# =========================
# 4. TOP COMMON AND SPARSE FRACTIONS
# =========================

top_common_fractions = fraction_summary.head(5).index.tolist()
top_sparse_fractions = fraction_summary.sort_values("zero_percentage", ascending=False).head(5).index.tolist()

print("Top 5 common fractions by mean share:")
print(top_common_fractions)

print("\nTop 5 sparse fractions by zero percentage:")
print(top_sparse_fractions)
```

```python
# =========================
# 5. TIME SERIES OF TOP COMMON FRACTIONS
# =========================

plt.figure(figsize=(14, 5))

for col in top_common_fractions:
    plt.plot(frac_df["Datum"], frac_df[col], label=f"Fraction {col}", alpha=0.8)

plt.title("Time Series of Top Common Fractions")
plt.xlabel("Date")
plt.ylabel("Fraction Share")
plt.legend()
plt.show()
```

```python
# =========================
# 6. 30-DAY ROLLING MEAN OF TOP COMMON FRACTIONS
# =========================

plt.figure(figsize=(14, 5))

for col in top_common_fractions:
    rolling_col = frac_df[col].rolling(window=30, min_periods=1).mean()
    plt.plot(frac_df["Datum"], rolling_col, label=f"Fraction {col}")

plt.title("30-Day Rolling Mean of Top Common Fractions")
plt.xlabel("Date")
plt.ylabel("Rolling Mean Fraction Share")
plt.legend()
plt.show()
```

```python
# =========================
# 7. TIME SERIES OF SPARSE FRACTIONS
# =========================

plt.figure(figsize=(14, 5))

for col in top_sparse_fractions:
    plt.plot(frac_df["Datum"], frac_df[col], label=f"Fraction {col}", alpha=0.8)

plt.title("Time Series of Sparse Fractions")
plt.xlabel("Date")
plt.ylabel("Fraction Share")
plt.legend()
plt.show()
```

```python
# =========================
# 8. YEARLY MEAN FRACTIONS
# =========================

yearly_fraction_means = frac_df.groupby("year")[fraction_cols].mean()

print("===== YEARLY MEAN FRACTIONS =====")
display(yearly_fraction_means)

plt.figure(figsize=(12, 5))
plt.imshow(yearly_fraction_means.T, aspect="auto")
plt.colorbar(label="Mean fraction share")
plt.title("Heatmap: Mean Fraction Share by Year")
plt.xlabel("Year")
plt.ylabel("Fraction Category")
plt.xticks(
    ticks=np.arange(len(yearly_fraction_means.index)),
    labels=yearly_fraction_means.index
)
plt.yticks(
    ticks=np.arange(len(fraction_cols)),
    labels=fraction_cols
)
plt.show()
```

```python
# =========================
# 9. MONTHLY MEAN FRACTIONS ACROSS ALL YEARS
# =========================

monthly_fraction_means = frac_df.groupby("month")[fraction_cols].mean()

print("===== MONTHLY MEAN FRACTIONS ACROSS ALL YEARS =====")
display(monthly_fraction_means)

plt.figure(figsize=(12, 5))
plt.imshow(monthly_fraction_means.T, aspect="auto")
plt.colorbar(label="Mean fraction share")
plt.title("Heatmap: Mean Fraction Share by Month")
plt.xlabel("Month")
plt.ylabel("Fraction Category")
plt.xticks(
    ticks=np.arange(len(monthly_fraction_means.index)),
    labels=monthly_fraction_means.index
)
plt.yticks(
    ticks=np.arange(len(fraction_cols)),
    labels=fraction_cols
)
plt.show()
```

```python
# ============================================================
# DATA MINING CUP - EDA PLOTS FOR PRESENTATION
# Generates all plots needed for Beamer slides
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# ------------------------------------------------------------
# 0. Output folder
# ------------------------------------------------------------

OUTDIR = Path("eda_plots")
OUTDIR.mkdir(exist_ok=True)

print("Plots will be saved in:", OUTDIR.resolve())


# ------------------------------------------------------------
# 1. Load data
# ------------------------------------------------------------

input_df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/time series/eingangsbandwaage_studenten.csv")
fraction_df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/time series/ballengewichte_studenten.csv")

input_df["Zeitstempel"] = pd.to_datetime(input_df["Zeitstempel"])
fraction_df["Datum"] = pd.to_datetime(fraction_df["Datum"])

fraction_cols = [col for col in fraction_df.columns if col != "Datum"]

# Rename for joining
input_join = input_df.rename(columns={"Zeitstempel": "date"}).copy()
fraction_join = fraction_df.rename(columns={"Datum": "date"}).copy()

# Inner and outer joins
inner_joined = pd.merge(input_join, fraction_join, on="date", how="inner")

outer_joined = pd.merge(
    input_join,
    fraction_join,
    on="date",
    how="outer",
    indicator=True
).sort_values("date").reset_index(drop=True)

print("Input shape:", input_df.shape)
print("Fraction shape:", fraction_df.shape)
print("Inner joined shape:", inner_joined.shape)
print("Outer joined shape:", outer_joined.shape)


# ------------------------------------------------------------
# 2. Helper function for saving figures
# ------------------------------------------------------------

def savefig(filename):
    path = OUTDIR / filename
    plt.savefig(path, dpi=200, bbox_inches="tight")
    plt.show()
    print("Saved:", path)


# ============================================================
# PLOT 1: Dataset overview and join summary
# File: dataset_join_summary.png
# ============================================================

# Missing calendar dates
input_full_dates = pd.date_range(
    start=input_df["Zeitstempel"].min(),
    end=input_df["Zeitstempel"].max(),
    freq="D"
)

fraction_full_dates = pd.date_range(
    start=fraction_df["Datum"].min(),
    end=fraction_df["Datum"].max(),
    freq="D"
)

missing_input_dates = input_full_dates.difference(input_df["Zeitstempel"])
missing_fraction_dates = fraction_full_dates.difference(fraction_df["Datum"])

overview_table = pd.DataFrame({
    "Dataset": ["Incoming mass", "Material fractions"],
    "Rows": [input_df.shape[0], fraction_df.shape[0]],
    "Columns": [input_df.shape[1], fraction_df.shape[1]],
    "Start date": [
        input_df["Zeitstempel"].min().date(),
        fraction_df["Datum"].min().date()
    ],
    "End date": [
        input_df["Zeitstempel"].max().date(),
        fraction_df["Datum"].max().date()
    ],
    "Null values": [
        input_df.isna().sum().sum(),
        fraction_df.isna().sum().sum()
    ],
    "Duplicate dates": [
        input_df["Zeitstempel"].duplicated().sum(),
        fraction_df["Datum"].duplicated().sum()
    ],
    "Missing calendar dates": [
        len(missing_input_dates),
        len(missing_fraction_dates)
    ]
})

join_counts = outer_joined["_merge"].value_counts().reset_index()
join_counts.columns = ["Join status", "Count"]

fig, axes = plt.subplots(2, 1, figsize=(13, 5.8))
fig.suptitle("Dataset Overview and Join Summary", fontsize=16, fontweight="bold")

axes[0].axis("off")
table1 = axes[0].table(
    cellText=overview_table.values,
    colLabels=overview_table.columns,
    loc="center",
    cellLoc="center"
)
table1.auto_set_font_size(False)
table1.set_fontsize(8.5)
table1.scale(1, 1.5)
axes[0].set_title("Dataset overview", fontsize=12, pad=8)

axes[1].axis("off")
table2 = axes[1].table(
    cellText=join_counts.values,
    colLabels=join_counts.columns,
    loc="center",
    cellLoc="center"
)
table2.auto_set_font_size(False)
table2.set_fontsize(10)
table2.scale(1, 1.5)
axes[1].set_title("Outer join status", fontsize=12, pad=8)

plt.tight_layout()
savefig("dataset_join_summary.png")


# ============================================================
# PLOT 2: Missing fraction dates by weekday
# File: missing_fraction_weekday.png
# ============================================================

fraction_start_date = fraction_df["Datum"].min()

left_only_dates = outer_joined[outer_joined["_merge"] == "left_only"].copy()

true_missing_fraction_dates = left_only_dates[
    left_only_dates["date"] >= fraction_start_date
].copy()

true_missing_fraction_dates["weekday"] = true_missing_fraction_dates["date"].dt.day_name()

weekday_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]

missing_weekday_counts = (
    true_missing_fraction_dates["weekday"]
    .value_counts()
    .reindex(weekday_order)
    .fillna(0)
    .astype(int)
)

mass_status = pd.Series({
    "Masse = 0": (true_missing_fraction_dates["Masse"] == 0).sum(),
    "Masse > 0": (true_missing_fraction_dates["Masse"] > 0).sum(),
    "Masse < 0": (true_missing_fraction_dates["Masse"] < 0).sum()
})

fig, axes = plt.subplots(1, 2, figsize=(13, 4.8))
fig.suptitle("Missing Fraction Dates Are Operationally Structured", fontsize=15, fontweight="bold")

axes[0].bar(missing_weekday_counts.index, missing_weekday_counts.values)
axes[0].set_title("True missing fraction dates by weekday")
axes[0].set_ylabel("Count")
axes[0].tick_params(axis="x", rotation=45)

axes[1].bar(mass_status.index, mass_status.values)
axes[1].set_title("Masse status on missing fraction dates")
axes[1].set_ylabel("Count")
axes[1].tick_params(axis="x", rotation=20)

plt.tight_layout()
savefig("missing_fraction_weekday.png")


# ============================================================
# PLOT 3: Weekly distribution of Masse
# File: masse_weekday_zero.png
# ============================================================

mass_df = input_df.copy()
mass_df["weekday"] = mass_df["Zeitstempel"].dt.day_name()
mass_df["is_zero_mass"] = mass_df["Masse"] == 0

weekday_mass_summary = (
    mass_df
    .groupby("weekday")["Masse"]
    .agg(["count", "mean", "median"])
    .reindex(weekday_order)
)

zero_by_weekday = (
    mass_df
    .groupby("weekday")["is_zero_mass"]
    .mean()
    .reindex(weekday_order) * 100
)

fig, axes = plt.subplots(1, 2, figsize=(13, 4.8))
fig.suptitle("Weekly Operating Pattern in Input Mass", fontsize=15, fontweight="bold")

axes[0].bar(weekday_mass_summary.index, weekday_mass_summary["mean"])
axes[0].set_title("Mean input mass by weekday")
axes[0].set_ylabel("Mean normalized Masse")
axes[0].tick_params(axis="x", rotation=45)

axes[1].bar(zero_by_weekday.index, zero_by_weekday.values)
axes[1].set_title("Zero-mass percentage by weekday")
axes[1].set_ylabel("Zero-mass days (%)")
axes[1].tick_params(axis="x", rotation=45)

plt.tight_layout()
savefig("masse_weekday_zero.png")


# ============================================================
# PLOT 4: Masse trend over time with rolling mean
# File: masse_rolling_trend.png
# ============================================================

mass_trend = input_df.copy().sort_values("Zeitstempel")
mass_trend["rolling_7"] = mass_trend["Masse"].rolling(window=7, min_periods=1).mean()
mass_trend["rolling_30"] = mass_trend["Masse"].rolling(window=30, min_periods=1).mean()

plt.figure(figsize=(13, 4.8))
plt.plot(mass_trend["Zeitstempel"], mass_trend["Masse"], alpha=0.35, label="Daily Masse")
plt.plot(mass_trend["Zeitstempel"], mass_trend["rolling_7"], label="7-day rolling mean")
plt.plot(mass_trend["Zeitstempel"], mass_trend["rolling_30"], label="30-day rolling mean")

plt.title("Input Mass Over Time", fontsize=15, fontweight="bold")
plt.xlabel("Date")
plt.ylabel("Normalized Masse")
plt.legend()
plt.grid(alpha=0.25)

savefig("masse_rolling_trend.png")


# ============================================================
# PLOT 5: Material-wise average share
# File: material_average_share.png
# ============================================================

fraction_summary = pd.DataFrame({
    "mean": fraction_df[fraction_cols].mean(),
    "median": fraction_df[fraction_cols].median(),
    "std": fraction_df[fraction_cols].std(),
    "min": fraction_df[fraction_cols].min(),
    "max": fraction_df[fraction_cols].max(),
    "zero_count": (fraction_df[fraction_cols] == 0).sum(),
    "zero_percentage": (fraction_df[fraction_cols] == 0).mean() * 100
})

fraction_summary_sorted = fraction_summary.sort_values("mean", ascending=False)

plt.figure(figsize=(11, 4.8))
plt.bar(fraction_summary_sorted.index, fraction_summary_sorted["mean"])
plt.title("Average Share of Each Material Fraction", fontsize=15, fontweight="bold")
plt.xlabel("Material fraction")
plt.ylabel("Mean fraction share")
plt.grid(axis="y", alpha=0.25)

# Mark Material 1
if "1" in fraction_summary_sorted.index:
    x_pos = list(fraction_summary_sorted.index).index("1")
    y_val = fraction_summary_sorted.loc["1", "mean"]
    plt.text(x_pos, y_val + 0.005, "Material 1\nhighest", ha="center", fontsize=9)

savefig("material_average_share.png")


# ============================================================
# PLOT 6: Material-wise zero percentage
# File: material_zero_percentage.png
# ============================================================

fraction_zero_sorted = fraction_summary.sort_values("zero_percentage", ascending=False)

plt.figure(figsize=(11, 4.8))
plt.bar(fraction_zero_sorted.index, fraction_zero_sorted["zero_percentage"])
plt.title("Zero Percentage of Each Material Fraction", fontsize=15, fontweight="bold")
plt.xlabel("Material fraction")
plt.ylabel("Zero observations (%)")
plt.grid(axis="y", alpha=0.25)

savefig("material_zero_percentage.png")


# ============================================================
# PLOT 7: Correlation between material fractions
# File: fraction_correlation_heatmap.png
# ============================================================

fraction_corr = fraction_df[fraction_cols].corr()

plt.figure(figsize=(8, 6.5))
plt.imshow(fraction_corr, vmin=-1, vmax=1, aspect="auto")
plt.colorbar(label="Correlation")
plt.title("Correlation Heatmap Between Material Fractions", fontsize=14, fontweight="bold")
plt.xlabel("Material fraction")
plt.ylabel("Material fraction")
plt.xticks(ticks=np.arange(len(fraction_cols)), labels=fraction_cols)
plt.yticks(ticks=np.arange(len(fraction_cols)), labels=fraction_cols)

savefig("fraction_correlation_heatmap.png")


# ============================================================
# PLOT 8: Correlation between Masse and material fractions
# File: masse_fraction_correlation.png
# ============================================================

mf_df = inner_joined.copy()
mf_df = mf_df[mf_df["Masse"] >= 0].copy()

mf_fraction_cols = [col for col in fraction_cols if col in mf_df.columns]

masse_fraction_corr = (
    mf_df[["Masse"] + mf_fraction_cols]
    .corr()["Masse"]
    .drop("Masse")
    .sort_values()
)

# Mass groups for context
mf_df["mass_group"] = pd.cut(
    mf_df["Masse"],
    bins=[-0.001, 0.000001, 0.33, 0.66, 1.0],
    labels=["zero", "low", "medium", "high"]
)

mean_fractions_by_mass_group = (
    mf_df
    .groupby("mass_group")[mf_fraction_cols]
    .mean()
)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle("Relationship Between Input Mass and Material Composition", fontsize=15, fontweight="bold")

axes[0].bar(masse_fraction_corr.index, masse_fraction_corr.values)
axes[0].axhline(0, linewidth=1)
axes[0].set_title("Correlation: Masse vs material fractions")
axes[0].set_xlabel("Material fraction")
axes[0].set_ylabel("Correlation")
axes[0].tick_params(axis="x", rotation=0)

im = axes[1].imshow(mean_fractions_by_mass_group.T, aspect="auto")
axes[1].set_title("Mean fraction share by mass group")
axes[1].set_xlabel("Mass group")
axes[1].set_ylabel("Material fraction")
axes[1].set_xticks(np.arange(len(mean_fractions_by_mass_group.index)))
axes[1].set_xticklabels(mean_fractions_by_mass_group.index, rotation=30)
axes[1].set_yticks(np.arange(len(mf_fraction_cols)))
axes[1].set_yticklabels(mf_fraction_cols)
fig.colorbar(im, ax=axes[1], label="Mean share")

plt.tight_layout()
savefig("masse_fraction_correlation.png")


# ============================================================
# PLOT 9: Trend over years for material fractions
# File: yearly_fraction_heatmap.png
# ============================================================

yearly_fraction_df = fraction_df.copy()
yearly_fraction_df["year"] = yearly_fraction_df["Datum"].dt.year

yearly_fraction_means = yearly_fraction_df.groupby("year")[fraction_cols].mean()

plt.figure(figsize=(10.5, 5.8))
plt.imshow(yearly_fraction_means.T, aspect="auto")
plt.colorbar(label="Mean fraction share")
plt.title("Mean Material Fraction Share by Year", fontsize=15, fontweight="bold")
plt.xlabel("Year")
plt.ylabel("Material fraction")
plt.xticks(
    ticks=np.arange(len(yearly_fraction_means.index)),
    labels=yearly_fraction_means.index
)
plt.yticks(
    ticks=np.arange(len(fraction_cols)),
    labels=fraction_cols
)

savefig("yearly_fraction_heatmap.png")


# ============================================================
# PLOT 10: Why Material 1? Trend and rolling mean
# File: material1_rolling_trend.png
# ============================================================

target_material = "1"

mat1_df = fraction_df[["Datum", target_material]].copy()
mat1_df = mat1_df.sort_values("Datum")
mat1_df["rolling_30"] = mat1_df[target_material].rolling(window=30, min_periods=1).mean()
mat1_df["rolling_90"] = mat1_df[target_material].rolling(window=90, min_periods=1).mean()

mat1_mean = mat1_df[target_material].mean()
mat1_zero_pct = (mat1_df[target_material] == 0).mean() * 100

plt.figure(figsize=(13, 4.8))
plt.plot(mat1_df["Datum"], mat1_df[target_material], alpha=0.35, label="Material 1 daily fraction")
plt.plot(mat1_df["Datum"], mat1_df["rolling_30"], label="30-day rolling mean")
plt.plot(mat1_df["Datum"], mat1_df["rolling_90"], label="90-day rolling mean")

plt.axhline(mat1_mean, linestyle="--", linewidth=1, label=f"Mean = {mat1_mean:.3f}")

plt.title("Material 1 Over Time: Highest Average Fraction", fontsize=15, fontweight="bold")
plt.xlabel("Date")
plt.ylabel("Fraction share")
plt.legend()
plt.grid(alpha=0.25)

text = f"Mean share: {mat1_mean:.3f}\nZero percentage: {mat1_zero_pct:.2f}%"
plt.text(
    0.01, 0.95, text,
    transform=plt.gca().transAxes,
    verticalalignment="top",
    bbox=dict(boxstyle="round", alpha=0.15)
)

savefig("material1_rolling_trend.png")


# ============================================================
# OPTIONAL: Export summary tables as CSV
# ============================================================

overview_table.to_csv(OUTDIR / "dataset_overview_table.csv", index=False)
join_counts.to_csv(OUTDIR / "join_counts.csv", index=False)
fraction_summary.to_csv(OUTDIR / "fraction_summary.csv")
masse_fraction_corr.to_csv(OUTDIR / "masse_fraction_correlation.csv")
yearly_fraction_means.to_csv(OUTDIR / "yearly_fraction_means.csv")

print("\nAll plots saved in:", OUTDIR.resolve())
print("Files created:")
for f in sorted(OUTDIR.glob("*.png")):
    print("-", f.name)
```
